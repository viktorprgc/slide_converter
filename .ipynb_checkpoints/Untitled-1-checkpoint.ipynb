```python
import subprocess
import os
import shutil
import sys
import copy
from pathlib import Path
```

```python
def norm_str(s):
    import unicodedata
    if s is None:
        return None
    
    s = unicodedata.normalize("NFKD", s)
    s = "".join(c for c in s if not unicodedata.combining(c))
    s = s.lower().strip()
    s = s.replace(" ", "_")
    return s
```

```python
class imageConv:
    def __init__(
        self,
        in_file: Path,
        out_path: Path,
    ):
        self.in_file = Path(in_file)
        self.out_path = Path(out_path)
        self.basecmd = [ 'conda', 'run', '-n', 'remote' ]
        self.fname = norm_str(in_file.stem)
        self.raw_out_path = (self.out_path / f"{self.fname}_raw")
        self.tiff_out_file = (self.out_path / f"{self.fname}.tiff")

    def conv2raw(
        self,
        compression="blosc",
        cname="zstd",
        clevel=3,
        max_tiles=128
    ):
        image_file = self.in_file.resolve()
        raw_out_path = self.raw_out_path.resolve()

        if raw_out_path.exists():
            try:
                shutil.rmtree(raw_out_path)
            except OSError as e:
                print(f"Error: {raw_out_path} : {e.strerror}")

        cmd = self.basecmd.copy()

        arglist = [
            'bioformats2raw',
            f'{str(image_file)}',
            f'{str(raw_out_path)}',
            "-p",
            "--compression",
            compression,
            "--compression-properties",
            f"cname={cname}",
            "--compression-properties",
            f"clevel={clevel}",
            "--max-cached-tiles",
            str(max_tiles),
        ]

        for arg in arglist:
            cmd.append(arg)
        
        result = subprocess.run(cmd, check=True)

        return result
    
    def raw2tiff(
        self,
        compression="JPEG_2000_LOSSY",
        rgb=True,
        max_workers=8,
        quality=0.02
    ):
        assert self.raw_out_path.is_dir()

        tile_dir = self.raw_out_path.resolve()
        out_file = self.tiff_out_file.resolve()

        if out_file.exists():
            try:
                out_file.unlink()
            except OSError as e:
                print(f"Error: {out_file} : {e.strerror}")

        cmd = self.basecmd.copy()

        arglist = [
            'raw2ometiff',
            f'{str(tile_dir)}',
            f'{str(out_file)}',
            "-p",
            "--compression",
            compression,
            "--max_workers",
            str(max_workers),
            "--quality",
            str(quality),
        ]

        if rgb:
            arglist.append("--rgb")

        for arg in arglist:
            cmd.append(arg)

        result = subprocess.run(cmd, check=True)

        return result
```

```python
test_file = Path('t/lam/Patologia Geral/80 Apoptose.mrxs')
test_out = Path('t/converter/')

assert test_file.exists()
assert test_out.exists()
```

```python
test = imageConv(test_file, test_out)
print(test.in_file)
```
